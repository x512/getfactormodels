## CLI

Example notebook showing how to use `getfactormodels` from the command-line.

```python
%%bash
getfactormodels --version
```

#### Download a model

Use the `--model`/`-m` parameter with a model name, and optionally specify a frequency with `--frequency/-f`.

- By default, monthly `m` data is returned. One of `d`, `w`, `m`, `q`, `y` or `w2w` (q-factors only).

```python
%%bash
getfactormodels --model ff3 --frequency m
```

- Use the `--list-models` command to see available models and frequencies.

```python
%%bash 
getfactormodels --list-models
```

Specify a date range using `-s`/`--start` and `-e`/`--end`.
- Date inputs accepts `YYYY`, `YYYY-MM`, or `YYYY-MM-DD`. All dates are inclusive.

```python
%%bash
getfactormodels --model ff5 --frequency d --start 2000 --end 2010-12-31
```

- You can save to file with the `-o`/`--output` param. Accepts a filename or path with extension. Supports .csv, .txt, .parquet, .ipc, .md.

```python
%%bash
getfactormodels -m icr -f d -s 2010-01 -e 2020 --output "~/icr_factors.md"
```

- Use `--quiet`/`-q` to supress all output to console.

- Get multiple models of the same frequency

```python
%%bash
getfactormodels -m ff3 liq -f m -s 2000 -e 2010
```

- Extract or drop factor(s) from a model with `--drop`/`-d` and `--extract`/`-x`

```python
%%bash
getfactormodels -m carhart -f d -x MOM
```

```python
%%bash
getfactormodels -m ff6 -d y -d RF HML
```

- Some models (Fama-French and AQR models) have regional data. Use `-r`-/`--region` with the models that support it.

```python
%%bash
#TODO: redo this output.
getfactormodels --list-regions
```

- For example, to get the monthly Betting Against Beta factors for Australia:

```python
%%bash
getfactormodels -m bab -f m --region aus --start 2020
```

- or the Fama-French 6-Factor model for Europe:

```python
%%bash
getfactormodels -m ff6 -f m -r europe
```

### Portfolios

Download portfolio data with the `--portfolio`/`-p` parameter. Supports portfolio sorts and industry returns from Kenneth French Data Library, and global-q.org's portfolios. AQR added soon.

* Fama-French industry returns:  5, 10, 12, 17, 30, 38, 48, 49.

```python
%%bash
getfactormodels --portfolio industry 5
# optionally, use --industry/-I
#getfactormodels -I 5
```

```python
%%bash
# Daily 30 industry returns
getfactormodels --portfolio industry 12 --frequency d
```

```python
%%bash
# 10 portfolios sorted on Book-to-Market (Monthly)
getfactormodels --portfolio bm -n 10
#getfactormodels -p bm --sort decile
```

```python
%%bash
# 5 portfolios sorted on accruals (ac)
getfactormodels -p ac -n quintile
#getfactormodels --portfolio ac --sort 5
```

```python
%%bash
# Bivariate: 6 portfolios sorted on size and OP
# --sort/-n accepts the sort (2x3) or portfolios (6), and --weights/-W accepts eithers value `vw` (default) or equal `ew`
getfactormodels --portfolio size op -n 6
getfactormodels -p op,size -n 5x5
```

```python
%%bash
#6 portfolios sorted on size and book-to-market
getfactormodels -p bm size -n 6
#getfactormodels --portfolio bm,size --sort 2x3
```

```python
%%bash
# 5x5 sort (25 portfolios) on size and market beta
getfactormodels -p size beta -n 5x5
#getfactormodels --portfolio size,beta --sort 25
```

* By default, portfolios contain value-weighted portfolios, `--weight`/`-W` accepts value-weighted `vw` or equal-weighted `ew`.

```python
%%bash
getfactormodels -p size,beta -n 5x5 --weights ew
```

* The Kenneth French Data library offers three 3-way sorts (2x4x4, 32 portfolios, monthly/annual only, div and ex div) too:

```python
%%bash
# -n/--sort is redundant here.
#getfactormodels -p size,bm,op -n 32

# order doesn't matter
#getfactormodels --portfolio bm,size,inv

getfactormodels -p size,inv,op --start 2000-01
```

- Access the global-q (q-factors) portfolio sorts:

```python
%%bash
# Global-q offers 2 portfolios, 2x3 (6, default), and 2x3x3 (18)
getfactormodels -p q -n 6 #note -n 6 is redundant here
```

```python
%%bash
getfactormodels --portfolio q -n 18
# You can explicitly specify params:
#getfactormodels -p size,ia,roe -n 18
```

Attach a portfolio to a model:

```python
%%bash
getfactormodels --model ff3 --frequency m --portfolio op -n 5 --start 2010
```

```python
%%bash
# Premium-Labour Consumption CAPM + Fama-French 3 Factors + 5 industry returns
getfactormodels --model pcapm ff3 --portfolio industry 5 -f m -s 2019 -e 2022
```

```python
%%bash 
# Fama-French 6-Factors + 18 q-global portfolios
getfactormodels -m ff6 -f d -p q -n 18
```

```python
%%bash
#this just...quietly adds quarterly q-factors to 6 q portfolios
getfactormodels -m q -f q -p q -q
```
